```python
class Data_portfolio(Dataset):
    def __init__(self, flag, input_len):
        super().__init__()
        self.flag = flag
        self.input_len = input_len
        self.__read_data__()

    def __read_data__(self):
        df_all = pd.read_parquet('./data/df_all.parquet')
        self.split = int(0.7 * df_all.shape[0])
        if self.flag == 'train':
            self.data = df_all[:self.split]
            self.data = np.array(self.data)
        elif self.flag == 'test':
            self.data = df_all[self.split:]
    def __getitem__(self, index):  
        seq_x =self.data[index:index+self.input_len]
        seq_y = self.data[index+self.input_len:index+self.input_len+1,2::3]
        seq_x = torch.tensor(seq_x)
        seq_y = torch.tensor(seq_y)
        return seq_x, seq_y
    
    def __len__(self):
        return self.data.shape[0]-self.input_len
            
```

```python
import torch.nn as nn
class ResBlock(nn.Module):
    def __init__(self, seq_len, pred_len, d_model, dropout, enc_in):
        super(ResBlock, self).__init__()

        self.temporal = nn.Sequential(
            nn.Linear(seq_len, d_model),
            nn.ReLU(),
            nn.Linear(d_model, seq_len),
            nn.Dropout(dropout)
        )

        self.channel = nn.Sequential(
            nn.Linear(enc_in, d_model),
            nn.ReLU(),
            nn.Linear(d_model, enc_in),
            nn.Dropout(dropout)
        )

    def forward(self, x):
        # x: [B, L, D]
        x = x + self.temporal(x.transpose(1, 2)).transpose(1, 2)
        x = x + self.channel(x)
        return x


class Model(nn.Module):
    def __init__(self, seq_len, pred_len, d_model, dropout, enc_in, method, batch_size, revin = False):
        super(Model, self).__init__()
        self.layer = 3
        self.model = nn.ModuleList([ResBlock(seq_len, pred_len, d_model, dropout, enc_in)
                                    for _ in range(3)])
        # self.M = torch.tensor(pkl.load(open('projection_mat.pkl', 'rb'))).unsqueeze(0).repeat(batch_size,1,1).float()
        self.pred_len = pred_len
        self.seq_len = seq_len
        self.method = method

        self.projection = nn.Linear(self.seq_len, self.pred_len)
        self.projection2 = nn.Linear(enc_in, 7)
        self.last_activ = nn.Softmax(dim =1)
        self.revin = revin

    def forecast(self, x_enc):
 
        for i in range(self.layer):
           x_enc = self.model[i](x_enc)
        
        enc_out = self.projection(x_enc.transpose(1, 2)).transpose(1, 2)
        enc_out = self.projection2(enc_out)
        return enc_out

    def forward(self, x_enc):
        dec_out = self.forecast(x_enc).squeeze()
        dec_out = self.last_activ(dec_out)
        return dec_out
  
```

```python
class ReplayBuffer:
    def __init__(self, start_index, end_index, batch_size, is_permed, coin_number, sample_bias=1.0):
        """
        :param start_index: start index of the training set on the global data matrices
        :param end_index: end index of the training set on the global data matrices
        """
        self.__coin_number = coin_number
        self.__experiences = [Experience(i) for i in range(start_index, end_index)]
        self.__is_permed = is_permed
        # NOTE: in order to achieve the previous w feature
        self.__batch_size = batch_size
        self.__sample_bias = sample_bias

    def append_experience(self, state_index):
        self.__experiences.append(Experience(state_index))

    def __sample(self, start, end, bias):
        """
        @:param end: is excluded
        @:param bias: value in (0, 1)
        """
        # TODO: deal with the case when bias is 0
        ran = np.random.geometric(bias)
        while ran > end - start:
            ran = np.random.geometric(bias)
        result = end - ran
        return result

    def next_experience_batch(self):
        # First get a start point randomly
        batch = []
        if self.__is_permed:
            for i in range(self.__batch_size):
                batch.append(self.__experiences[self.__sample(self.__experiences[0].state_index,
                                                              self.__experiences[-1].state_index,
                                                              self.__sample_bias)])
        else:
            batch_start = self.__sample(0, len(self.__experiences) - self.__batch_size,
                                        self.__sample_bias)
            batch = self.__experiences[batch_start:batch_start+self.__batch_size]
        return batch


class Experience:
    def __init__(self, state_index):
        self.state_index = int(state_index)
```
